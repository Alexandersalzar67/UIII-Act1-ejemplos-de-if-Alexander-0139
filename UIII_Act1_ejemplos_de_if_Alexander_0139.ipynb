### ejmplo de if, if else, if elif

```python
#Este ejemplo utiliza operadores relacionales y lógicos para verificar si un número está dentro de un intervalo específico

numero = 15
if numero >= 10 and numero <= 20:
    print("El número está entre 10 y 20.")
```

```python
# Ejemplo de if-else (Verificación de año bisiesto)
# Aquí se combinan operadores aritméticos (módulo), relacionales (igualdad y diferencia) y lógicos para determinar una propiedad matemática compleja

anio = 2024
if (anio % 4 == 0 and anio % 100 != 0) or (anio % 400 == 0):
    print(f"{anio} es un año bisiesto.")
else:
    print(f"{anio} no es un año bisiesto.")
```

```python
 # Ejemplo de if-elif-else (Categorización por puntaje)
# Este esquema es ideal cuando existen más de dos caminos posibles. Utiliza operadores relacionales para clasificar un valor en diferentes escalas

puntaje = 85
if puntaje >= 90:
    print("Excelente")
elif puntaje >= 80:
    print("Muy bueno")
elif puntaje >= 70:
    print("Bueno")
else:
    print("Necesita mejorar")
```

```python
# Ejemplo combinado (Cálculo de descuentos)
# Este código utiliza operadores aritméticos para aplicar un cálculo sobre una variable y operadores relacionales para decidir el porcentaje de descuento
cantidad = 12
precio_unidad = 100
subtotal = cantidad * precio_unidad

if cantidad >= 10:
    total = subtotal * 0.80  # 20% de descuento
else:
    total = subtotal

print(f"El total a pagar es: {total}")
```
